Build a User Configuration Manager

```python
def add_setting(settings, new_setting):
    key, value = new_setting
    key = key.lower()
    value = value.lower()

    if key in settings.keys():
        return(f"Setting '{key}' already exists! Cannot add a new setting with this name.")
    else:
        settings[key] = value
        return(f"Setting '{key}' added with value '{value}' successfully!")

def update_setting(settings, new_setting):
    key, value = new_setting
    value = value.lower()

    if key in settings.keys():
        settings[key] = value
        return(f"Setting '{key}' updated to '{value}' successfully!")
    else:
        return(f"Setting '{key}' does not exist! Cannot update a non-existing setting.")

def delete_setting(settings, key):
    key = key.lower()
    if key in settings.keys():
        settings.pop(key)
        return(f"Setting '{key}' deleted successfully!")
    else:
        return(f"Setting not found!")

def view_settings(settings):
    string = "Current User Settings:\n"
    if not settings:
        return("No settings available.")
    else:
        for key_value_pair in settings.items():
            key = key_value_pair[0].title()
            value = key_value_pair[1]
            string += f"{key}: {value}\n"
    return(print(string))
```

```python
view_settings({'theme': 'dark', 'notifications': 'enabled', 'volume': 'high'})
```
